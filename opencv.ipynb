```python
# OpenCV 라이브러리 불러오기
# 영상(카메라), 이미지 처리에 사용되는 핵심 라이브러리
import cv2 as cv

# 시스템 관련 기능 라이브러리
# 에러 발생 시 프로그램을 종료하기 위해 사용
import sys


# 카메라(웹캠)를 여는 코드
# 0 : 기본 웹캠 (노트북 내장 카메라)
# cv.CAP_DSHOW : Windows에서 카메라 인식 오류를 줄이기 위한 옵션
cap=cv.VideoCapture(0,cv.CAP_DSHOW)



# 카메라가 정상적으로 열렸는지 확인
# 열리지 않았다면(False) → 카메라 사용 불가
if not cap.isOpened():
    # 에러 메시지를 출력하고 프로그램을 즉시 종료
    sys.exit("카메라 연결 실패")

# 카메라에서 영상을 계속 받아오기 위한 무한 반복문
while True:
    # 카메라로부터 프레임(영상 한 장)을 읽어옴
    # ret  : 프레임을 정상적으로 읽었는지 여부 (True / False)
    # frame: 실제 영상 데이터 (이미지 한 장)
    ret,frame=cap.read()

     # 프레임을 읽지 못했을 경우
    if not ret:
        # 에러 메시지를 출력
        print("프레임 획득에 실패하여 루프를 나갑니다.")
         # while 반복문 종료
        break

    # 현재 프레임(영상 한 장)을 화면에 출력
    # 'Video Display'는 창의 제목
    cv.imshow('Video Display',frame)

    # 키 입력을 아주 짧게(1ms) 기다림
    # 영상이 끊기지 않도록 반드시 필요
    key=cv.waitKey(1)
    # 만약 사용자가 'q' 키를 눌렀다면
    # ord('q') : 문자 'q'를 숫자 코드로 변환
    if key ==ord('q'):
        # while 반복문 종료
        break

# 카메라 사용이 끝났으므로 해제 (반납)
# 이 코드를 안 쓰면 다른 프로그램에서 카메라 사용 불가
cap.release()

# OpenCV로 열었던 모든 창을 닫음
# 프로그램 종료 전 정리 작업
cv.destroyAllWindows()
```

```python
# OpenCV 라이브러리 불러오기
# 카메라, 영상, 이미지 처리를 담당
import cv2 as cv

# NumPy 라이브러리 불러오기
# 이미지(frame)는 NumPy 배열 형태이기 때문에 함께 사용됨
import numpy as np

# 시스템 관련 기능 라이브러리
# 에러 발생 시 프로그램을 종료하기 위해 사용
import sys



# 웹캠(카메라)을 여는 코드
# 0 : 기본 웹캠
# cv.CAP_DSHOW : Windows에서 카메라 인식 오류를 줄이기 위한 옵션
cap=cv.VideoCapture(0,cv.CAP_DSHOW)

# 카메라가 정상적으로 열렸는지 확인
# 열리지 않았으면 False 반환
if not cap.isOpened():
     # 에러 메시지를 출력하고 프로그램을 즉시 종료
    sys.exit("카메라 연결 실패")

# 캡처한 프레임(이미지)을 저장할 빈 리스트 생성
# 'c' 키를 누를 때마다 현재 화면(frame)이 여기에 저장될 예정
frames=[]
# 카메라에서 영상을 계속 받아오기 위한 무한 반복문
while True:
    # 카메라로부터 프레임(영상의 한 장)을 읽어옴
    # ret   : 프레임을 성공적으로 읽었는지 여부 (True / False)
    # frame : 실제 영상 데이터 (컬러 이미지, NumPy 배열)
    ret,frame=cap.read()
    # 프레임을 읽는 데 실패했다면
    if not ret:
         # 에러 메시지 출력
        print("프레임 획득에 실패하여 루프를 나갑니다.")
        # while 반복문 종료
        break

    # 현재 프레임을 화면에 출력
    # 'Video Display'는 창 제목
    cv.imshow('Video Display',frame)

    # 키 입력을 아주 짧게(1ms) 기다림
    # 영상이 끊기지 않도록 반드시 필요
    key=cv.waitKey(1)

    # 만약 사용자가 키보드에서 'c' 키를 눌렀다면
    # ord('c') : 문자 'c'를 키보드 숫자 코드로 변환
    # c를 눌렀다면 캡쳐기능
    if key==ord('c'):
         # 👉  "현재 프레임 저장" 같은 코드
        frames.append(frame)
    #만약 사용자가키보드에서 'q'키를 눌렀다면
    # 나감
    elif key==ord('q'):
        break
# 카메라 사용이 끝났으므로 해제 (반납)
# 이 코드를 안 쓰면 다른 프로그램에서 카메라 사용 불가
cap.release()

# OpenCV로 열었던 모든 창을 닫음
# 프로그램 종료 전 정리 작업
cv.destroyAllWindows()
```

```python
# frames 리스트에 저장된 이미지가 하나라도 있는지 확인
# (아무 이미지도 없으면 아래 코드를 실행하면 안 됨)
if len(frames) >0:
    # 첫 번째로 캡처한 이미지를 기준 이미지로 사용
    # 이후 이 이미지에 다른 이미지들을 계속 이어 붙일 예정
    imgs=frames[0]
    # 두 번째 이미지부터 반복 시작
    # min(3, len(frames)) :
    #   - 최대 3장까지만 사용
    #   - frames에 이미지가 3장보다 적으면 그 개수만큼만 사용
    for i in range(1,min(3,len(frames))):
        # 현재까지 합쳐진 이미지(imgs)와
        # frames[i] 이미지를 가로 방향으로 이어 붙임
        # np.hstack() : horizontal stack (가로로 붙이기)
        imgs=np.hstack((imgs,frames[i]))

    # 가로로 이어 붙인 최종 이미지를 화면에 출력
    # 'collected Frames'는 창 제목
    cv.imshow('collected Frames',imgs)
    # 키 입력을 기다림
    # 아무 키나 누를 때까지 화면이 유지됨
    cv.waitKey()
    # OpenCV로 열었던 모든 창을 닫음
    cv.destroyAllWindows()
```

```python
# OpenCV 라이브러리 불러오기
# 이미지 읽기, 도형 그리기, 글자 출력 등에 사용
import cv2 as cv

# 시스템 관련 기능 라이브러리
# 오류 발생 시 프로그램을 종료하기 위해 사용
import sys

# 현재 폴더(./)에 있는 'girl_laughing.jpg' 이미지 파일을 읽어옴
# 성공하면 img에는 이미지 픽셀 데이터가 저장됨
# 실패하면 img에는 None이 저장됨
img=cv.imread('./girl_laughing.jpg')

# 만약 이미지 파일을 읽어오지 못했다면
# (파일이 없거나 경로가 잘못된 경우)
if img is None:
    # 에러 메시지를 출력하고 프로그램을 즉시 종료
    sys.exit("파일을 찾을 수 없습니다.")


# 이미지(img) 위에 사각형을 그림
# (830, 30)  : 사각형의 왼쪽 위 좌표 (x, y)
# (1000,200) : 사각형의 오른쪽 아래 좌표 (x, y)
# (0, 0, 255): 사각형 색상 (B, G, R) → 빨간색
# 2           : 사각형 선의 두께 (픽셀 단위)
cv.rectangle(img,(830,30),(1000,200),(0,0,255),2)

# 이미지(img) 위에 텍스트를 출력
# 'laugh'                     : 화면에 표시할 문자열
# (830, 24)                   : 글자가 시작되는 위치 (왼쪽 아래 기준)
# cv.FONT_HERSHEY_SIMPLEX     : 글꼴 종류 (기본 글꼴)
# 1                           : 글자 크기
# (255, 0, 0)                 : 글자 색상 (B, G, R) → 파란색
# 2                           : 글자 두께
cv.putText(img,'laugh',(830,24),cv.FONT_HERSHEY_SIMPLEX,1,(255,0,0),2)
```

```python
# 시스템 관련 기능을 사용하기 위한 라이브러리
# 오류 발생 시 프로그램을 종료할 때 사용
import sys

# OpenCV로 이미지 파일을 읽어옴
# './girl_laughing.jpg'는 현재 폴더에 있는 이미지 파일
img=cv.imread('./girl_laughing.jpg')



# 만약 이미지를 불러오지 못했다면
# (파일이 없거나 경로가 잘못된 경우)
if img is None:
    # 에러 메시지를 출력하고 프로그램을 즉시 종료
    sys.exit("파일을 찾을 수 없습니다.")


# 마우스 이벤트가 발생할 때 자동으로 호출될 함수 정의
# event : 발생한 마우스 이벤트 종류
# x, y  : 마우스 커서의 현재 좌표
# flags : 보조 키 상태 (Shift, Ctrl 등 / 여기서는 사용 안 함)
# param : 추가 전달 값 (여기서는 사용 안 함)
def draw(event,x,y,flags,param):
    # 전역 변수 선언 (이 코드에서는 실제로 사용하지 않음)
    # 없어도 동작하므로 생략 가능
    global ix,iy
     # 마우스 왼쪽 버튼을 눌렀을 때
    if event ==cv.EVENT_LBUTTONDOWN:
        # 마우스를 클릭한 위치(x, y)를 기준으로
        # 가로 200, 세로 200 크기의 빨간색 사각형을 그림
        cv.rectangle(img,# 사각형을 그릴 이미지
                     (x,y), # 사각형의 왼쪽 위 좌표
                     (x+200,y+200), # 사각형의 오른쪽 아래 좌표
                     (0,0,255),# 색상 (B, G, R) → 빨간색
                     2) # 선 두께
        
     # 마우스 오른쪽 버튼을 눌렀을 때    
    elif event ==cv.EVENT_RBUTTONDOWN:
         # 클릭한 위치를 기준으로
        # 가로 100, 세로 100 크기의 파란색 사각형을 그림
        cv.rectangle(img,
                     (x,y),
                     (x+100,y+100),
                     (255,0,0),# 색상 (B, G, R) → 파란색
                     2)
    # 사각형이 그려진 이미지를 다시 화면에 출력
    # (이 줄이 없으면 화면이 갱신되지 않음)
    cv.imshow('Drawing',img)


# 'Drawing'이라는 이름의 창을 생성
cv.namedWindow('Drawing')
# 처음 이미지를 화면에 출력
cv.imshow('Drawing',img)
# 'Drawing' 창에서 마우스 이벤트가 발생하면
# draw 함수를 자동으로 실행하도록 연결
cv.setMouseCallback('Drawing',draw)

# 프로그램이 바로 종료되지 않도록 무한 반복

while(True):
    # 키 입력을 1ms 동안 기다림
    # 만약 'q' 키가 눌리면
    if cv.waitKey(1)==ord('q'):
        # 모든 OpenCV 창을 닫고
        cv.destroyAllWindows()
        # 반복문을 종료하여 프로그램 끝냄
        break
```

```python
# OpenCV 라이브러리 불러오기
# 이미지 처리, 도형 그리기, 마우스 이벤트 처리에 사용
import cv2 as cv

# 시스템 관련 기능 라이브러리
# 오류 발생 시 프로그램을 종료하기 위해 사용
import sys


# 현재 폴더(./)에 있는 'soccer.jpg' 이미지 파일을 읽어옴
# 성공하면 img에는 이미지 픽셀 데이터가 저장됨
# 실패하면 img에는 None이 저장됨
img=cv.imread('./soccer.jpg')


# 만약 이미지를 불러오지 못했다면
# (파일이 없거나 경로가 잘못된 경우)
if img is None:
    # 에러 메시지를 출력하고 프로그램을 즉시 종료
    sys.exit('파일을 알수 없습니다.')

# 브러시(붓)의 크기 설정
# 값이 클수록 더 굵게 그림이 그려짐
BrushSiz=5

# 그릴 색상 설정 (B, G, R 순서)
# LColor : 마우스 왼쪽 버튼 색상 → 파란색
# RColor : 마우스 오른쪽 버튼 색상 → 빨간색
LColor,RColor=(255,0,0),(0,0,255)



# 마우스 이벤트가 발생할 때 자동으로 호출될 함수 정의
# event : 발생한 마우스 이벤트 종류
# x, y  : 마우스 커서의 현재 좌표
# flags : 마우스 버튼이 눌린 상태
# param : 추가로 전달되는 값 (여기서는 사용하지 않음)
def painting(event,x,y,flags,param):
    # 마우스 왼쪽 버튼을 눌렀을 때
    if event ==cv.EVENT_LBUTTONDOWN:
        # 클릭한 위치에 파란색 원을 그림
        # -1 : 원을 채워서 그림
        cv.circle(img,(x,y),BrushSiz,LColor,-1)
    # 마우스 오른쪽 버튼을 눌렀을 때
    elif event==cv.EVENT_RBUTTONDOWN:
         # 클릭한 위치에 빨간색 원을 그림
        cv.circle(img,(x,y),BrushSiz,RColor,-1)
    # 마우스를 움직이면서 왼쪽 버튼이 눌린 상태일 때
    # → 파란색으로 선처럼 계속 그림
    elif event==cv.EVENT_MOUSEMOVE and flags==cv.EVENT_FLAG_LBUTTON:
        cv.circle(img,(x,y),BrushSiz,LColor,-1)
    # 마우스를 움직이면서 오른쪽 버튼이 눌린 상태일 때
    # → 빨간색으로 선처럼 계속 그림
    elif event==cv.EVENT_MOUSEMOVE and flags==cv.EVENT_FLAG_RBUTTON:
        cv.circle(img,(x,y),BrushSiz,RColor,-1)
    # 그림이 그려질 때마다
    # 변경된 이미지를 화면에 다시 출력
    cv.imshow('Painting',img)

# 'Painting'이라는 이름의 창을 생성
cv.namedWindow('Painting')
# 처음 이미지를 화면에 출력
cv.imshow('Painting',img)

# 'Painting' 창에서 마우스 이벤트가 발생하면
# painting 함수를 자동으로 실행하도록 설정
cv.setMouseCallback('Painting',painting)

# 프로그램이 바로 종료되지 않도록 무한 반복
while(True):
    # 키 입력을 1ms 동안 기다림
    # 만약 'q' 키가 눌리면
    if cv.waitKey(1)==ord('q'):
        # 모든 OpenCV 창을 닫고
        cv.destroyAllWindows()
        # 반복문을 종료하여 프로그램을 끝냄
        break
```
