```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

import numpy as np
```

```python
# =========================
# 1️⃣ 데이터 불러오기
# =========================
# CSV 파일을 숫자 배열로 읽어옴
Data_set=np.loadtxt("./ThoraricSurgery3.csv",delimiter=",")
# 입력 데이터 (특성 16개)
X=Data_set[:,0:16]
# 정답 데이터 (0 또는 1)
y=Data_set[:,16]
```

```python
# =========================
# 2️⃣ 모델 생성
# =========================
# 레이어를 순서대로 쌓는 모델
model=Sequential()
# 첫 번째 은닉층
# 입력값 16개 → 뉴런 80개
# ReLU 활성화 함수 사용
model.add(Dense(80,input_dim=16,activation='relu'))
# 두 번째 은닉층
# 앞층의 출력을 자동으로 입력으로 사용
model.add(Dense(80,activation='relu'))
# 출력층
# 결과 1개 (이진 분류)
# sigmoid → 0~1 확률값 출력
model.add(Dense(1,activation='sigmoid'))
```

```python
# =========================
# 3️⃣ 모델 컴파일
# =========================
model.compile(loss='binary_crossentropy',# 이진 분류용 손실 함수
              optimizer='adam',  # 자동으로 잘 학습하는 최적화 방법
              metrics=['accuracy']) # 정확도 확인

# =========================
# 4️⃣ 모델 학습
# =========================
history=model.fit(X,# 입력 데이터, 정답
                  y,
                  epochs=80,  # 전체 데이터 80번 반복 학습
                  batch_size=16)# 16개씩 나눠서 학습
```
